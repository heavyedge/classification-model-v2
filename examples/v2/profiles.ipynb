# Profiles by each type

```python
from heavyedge import ProfileData
import matplotlib.pyplot as plt

with ProfileData("./profiles.h5") as data:
    Ys, _, _ = data[:]

titles = [
    "Type 0",
    "Type 1a",
    "Type 1b",
    "Type 2a",
    "Type 2b",
    "Type 3a",
    "Type 3b",
]

fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(12, 12))
axes_flat = axes.flatten()

for i, (Y, title) in enumerate(zip(Ys, titles)):
    ax = axes_flat[i]
    ax.plot(Y)
    ax.set_title(title)

for k in range(len(Ys), len(axes_flat)):
    axes_flat[k].set_visible(False)

plt.show()
```
